```python
import requests
import numpy as np
import json

# Define the API endpoint URL
url = 'http://127.0.0.1:5000/predict'

# Generate 48 random dummy values for the valid test
valid_48_values = np.random.uniform(20000, 40000, 48).tolist()

# Define our test cases (The Edge Cases)
test_cases = {
    "1. Valid Input (Happy Path)": {
        "payload": {"features": valid_48_values},
        "expected_status": 200
    },
    "2. Missing Key (Wrong dictionary key)": {
        "payload": {"wrong_key_name": valid_48_values},
        "expected_status": 400
    },
    "3. Too Few Values (Only 3 instead of 48)": {
        "payload": {"features": [25000, 26000, 27000]},
        "expected_status": 400
    },
    "4. Wrong Data Type (String instead of List)": {
        "payload": {"features": "This is a string, not a list of numbers"},
        "expected_status": 500 # Or 400 depending on how utils.py handles the crash
    },
    "5. Empty JSON": {
        "payload": {},
        "expected_status": 400
    }
}

print("STARTING API EDGE CASE TEST SUITE")
# Loop through and run each test
for test_name, test_data in test_cases.items():
    print(f"Running Test: {test_name}")
    
    # Send POST request
    response = requests.post(url, json=test_data["payload"])
    
    # Check results
    status_code = response.status_code
    response_json = response.json() if response.text else "No JSON response"
    
    # Print formatted output for your dissertation evidence
    if status_code == test_data["expected_status"] or (status_code in [400, 500] and test_data["expected_status"] in [400, 500]):
        print(f"  PASS (Status Code: {status_code})")
        print(f"  Response: {response_json}\n")
    else:
        print(f"  FAIL (Expected {test_data['expected_status']}, Got {status_code})")
        print(f"  Response: {response_json}\n")

print("TESTING COMPLETE")
```
